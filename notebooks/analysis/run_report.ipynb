# Run report

Presents **one recorded run** (a directory written by `beaversim.experiments.run`, see `beaversim/records.py`).
Set the parameters below and run all cells. Every figure comes from `beaversim.viz.single`; styling lives in
`beaversim.viz.theme` (palette + medium), so this notebook only chooses *what* to show.

```python
import sys
from pathlib import Path

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "beaversim").is_dir())
sys.path.insert(0, str(root))

from beaversim import viz
from beaversim.records import load_run
```

## Parameters

```python
RUN_DIR = root / "output/simulations/<experiment>/<condition>/seed_1"   # one run directory
MEDIUM = "chapter"      # "chapter", "paper" or "slides"
PALETTE = None          # None = default, or a path to a palette JSON from the designers
SAVE = False            # write each figure as PDF + PNG into FIG_DIR
FIG_DIR = root / "output/figures" / RUN_DIR.parent.name / RUN_DIR.name
```

```python
viz.use_theme(MEDIUM, PALETTE)
run = load_run(RUN_DIR)


def show(fig, name):
    """Display ``fig`` and optionally save it under FIG_DIR."""
    if SAVE:
        viz.save(fig, FIG_DIR / name)
    return fig
```

## What was simulated
Site, grid, duration, colony and the exact code version that produced the run.

```python
import pandas as pd
pd.Series(viz.single.run_card(run)).to_frame("run")
```

## 1. What changed in the landscape
Initial and final state of the landscape (negative = water depth, positive = vegetation quality), and their
difference: **loss** is vegetation harvested or terrain lowered, **gain** is regrowth or material stored.
Green dots mark home bases.

```python
show(viz.single.landscape_change(run), "landscape_change");
```

## 2. Where the agents travelled
Cumulative visits over the whole run from the agent trajectories, coloured by the dominant role in each cell
(intensity compressed with a power law to show faint trails). Water is drawn in grey for context.
Use `source="ledger"` to show the decaying stigmergic ledger instead, i.e. what agents sense at a given time.

```python
show(viz.single.trail_traffic(run), "trail_traffic");
```

## 3. How the landscape evolved
Evenly spaced snapshots of the landscape layer, sharing one colour scale.

```python
show(viz.single.snapshot_strip(run, n=5), "landscape_over_time");
```

## 4. Environment over time
Vegetation median with the 10–90% band across cells, and the share of open-water cells.

```python
show(viz.single.environment_over_time(run), "environment");
```

## 5. Swarm metrics
**Spatial diffusion** $\mathcal{D}$: cells ever visited, per role. **Canalization ratio** $\mathcal{C}$: share of
active traffic on the top 5% busiest cells (1-week moving average, first day hidden).

```python
show(viz.single.swarm_over_time(run, "diffusion"), "diffusion");
```

```python
show(viz.single.swarm_over_time(run, "canalization"), "canalization");
```

## Appendix — agent diagnostics
Carried load and distance to the current waypoint for every agent, coloured by role.

```python
show(viz.single.agent_diagnostics(run), "agents");
```
